```python
# ============================================================
# SKILLCRAFT TECHNOLOGY - TASK 01
# Population Distribution - Bar Chart
# ============================================================

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import requests

print("Libraries imported successfully!")


# ------------------------------------------------------------
# 1. Get World Bank Population Data
# ------------------------------------------------------------

countries = {
    "IND": "India",
    "CHN": "China",
    "USA": "United States",
    "BRA": "Brazil",
    "JPN": "Japan",
    "DEU": "Germany",
    "GBR": "United Kingdom",
    "FRA": "France",
    "CAN": "Canada",
    "AUS": "Australia"
}

year = 2022

data = []

for code, country in countries.items():

    url = f"https://api.worldbank.org/v2/country/{code}/indicator/SP.POP.TOTL?date={year}&format=json"

    response = requests.get(url)

    result = response.json()

    population = result[1][0]["value"]

    data.append({
        "Country": country,
        "Population": population
    })


# Create DataFrame
df = pd.DataFrame(data)


# ------------------------------------------------------------
# 2. Display Dataset
# ------------------------------------------------------------

print("\n========== POPULATION DATA ==========")

display(df)


# ------------------------------------------------------------
# 3. Check Dataset Information
# ------------------------------------------------------------

print("\n========== DATASET INFORMATION ==========")

print("Number of countries:", len(df))

print("\nMissing values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 4. Convert Population into Millions
# ------------------------------------------------------------

df["Population_Millions"] = df["Population"] / 1_000_000

print("\n========== POPULATION IN MILLIONS ==========")

display(df)


# ------------------------------------------------------------
# 5. Sort Population
# ------------------------------------------------------------

df_sorted = df.sort_values(
    by="Population",
    ascending=False
)

print("\n========== SORTED POPULATION ==========")

display(df_sorted)


# ------------------------------------------------------------
# 6. Create Bar Chart
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

plt.bar(
    df_sorted["Country"],
    df_sorted["Population_Millions"]
)

plt.title(
    "Population Distribution of Selected Countries - 2022",
    fontsize=16
)

plt.xlabel("Country", fontsize=12)

plt.ylabel("Population (Millions)", fontsize=12)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 7. Add Population Values to Chart
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

bars = plt.bar(
    df_sorted["Country"],
    df_sorted["Population_Millions"]
)

plt.title(
    "Population Distribution of Selected Countries - 2022",
    fontsize=16
)

plt.xlabel("Country")

plt.ylabel("Population (Millions)")

plt.xticks(rotation=45)

# Display values above bars
for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:.1f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 8. Basic Analysis
# ------------------------------------------------------------

largest_country = df.loc[
    df["Population"].idxmax()
]

smallest_country = df.loc[
    df["Population"].idxmin()
]

print("\n========== ANALYSIS ==========")

print(
    "Largest population:",
    largest_country["Country"],
    "-",
    round(largest_country["Population_Millions"], 2),
    "million"
)

print(
    "Smallest population:",
    smallest_country["Country"],
    "-",
    round(smallest_country["Population_Millions"], 2),
    "million"
)


# ------------------------------------------------------------
# 9. Final Conclusion
# ------------------------------------------------------------

print("\n========== CONCLUSION ==========")

print("""
The bar chart shows the population distribution of selected countries
in 2022.

China and India have the largest populations among the selected countries,
while Australia has the smallest population.

The visualization makes it easy to compare population sizes between
different countries.

This analysis demonstrates how a bar chart can be used to visualize
and compare a categorical variable (country) with a numerical variable
(population).
""")

print("TASK 01 COMPLETED SUCCESSFULLY!")
```

Visualize the demographic distribution of a population across continuous and categorical dimensions (age cohorts) based on data inspired by the United Nations World Population Prospects.

## 2. Methodology & Features

| Metric | Analysis Type | Description |
| :--- | :--- | :--- |

| **Population Base** | Macro Demographic Metric | Reflects an aggregate benchmark of **1.42 Billion**. |

## 3. Key Observations

- **Demographic Dividend**: The largest portion of the population (**57.0% / 807 Million**) falls within the productive working-age bracket (21 to 64 years).
- **Youth Dependency**: Youth aged 0 to 20 account for **36.1% (512 Million)**, representing substantial future labor potential alongside education and healthcare requirements.
- **Elderly Ratio**: The 65+ demographic comprises **6.9% (98 Million)**, indicating a comparatively low elderly dependency ratio relative to aging western economies.
